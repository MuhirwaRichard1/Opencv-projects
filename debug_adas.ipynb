```python

import os
import sys
import cv2

# Set path to include the ADAS src directory
sys.path.append(r"c:\Users\Robotic Muhirwa\Downloads\Tracking_Opencv\LaneCollisionADAS\src")

import config

print(f"Video path: {config.VIDEO_SOURCE}")
if not os.path.exists(config.VIDEO_SOURCE):
    print("Error: Video file not found!")
else:
    print("Video file exists.")

cap = cv2.VideoCapture(config.VIDEO_SOURCE)
if not cap.isOpened():
    print("Error: Could not open video capture.")
else:
    print("Video capture opened successfully.")
    ret, frame = cap.read()
    if not ret:
        print("Error: Could not read first frame.")
    else:
        print(f"Read first frame successfully. Shape: {frame.shape}")
    cap.release()
```
